# Step 10: Simple Linear Regression using Gradient Descent (From Scratch)

In this step, Gradient Descent is implemented from scratch without using Scikit-Learn's optimization algorithm. The objective is to iteratively update the slope and intercept by minimizing the Mean Squared Error (MSE) loss function until the model converges to the optimal solution.

## 10.1 : Preparing the Training Data

```python
# Convert DataFrame to NumPy arrays
X = X_train_simple.values.flatten()
Y = y_train_simple.values
```

## 10.2 : Initialize parameters 

```python
m = 0
c = 0

learning_rate = 1e-10
epochs = 1000

n = len(X)

loss_history = []
```

## 10.3 : Implement Gradient Descent

```python
for epoch in range(epochs):

    # Predicted values
    Y_pred = m * X + c

    # Mean Squared Error
    loss = (1/n) * np.sum((Y - Y_pred) ** 2)
    loss_history.append(loss)

    # Gradients
    dm = (-2/n) * np.sum(X * (Y - Y_pred))
    dc = (-2/n) * np.sum(Y - Y_pred)

    # Update parameters
    m = m - learning_rate * dm
    c = c - learning_rate * dc
```

## 10.4 : Final Parameters

```python
print("Gradient Descent Results")
print("-"*40)

print("Slope (m):", m)
print("Intercept (c):", c)
```

## 10.5 : Predictions

```python
y_pred_gd = m * X_test_simple.values.flatten() + c

print("First 5 Predictions:")
print(y_pred_gd[:5])
```

## 10.6 : evaluate the Model 

```python
mae_gd = mean_absolute_error(y_test_simple, y_pred_gd)
mse_gd = mean_squared_error(y_test_simple, y_pred_gd)
rmse_gd = np.sqrt(mse_gd)
r2_gd = r2_score(y_test_simple, y_pred_gd)

print("Mean Absolute Error:", mae_gd)
print("Mean Squared Error:", mse_gd)
print("Root Mean Squared Error:", rmse_gd)
print("R² Score:", r2_gd)
```

## 10.7 : Loss vs Epoch 

```python
plt.figure(figsize=(9,5))

plt.plot(
    loss_history,
    color="#7C3AED",
    linewidth=2.5
)

plt.title("Gradient Descent Loss vs Epoch",
          fontsize=16,
          fontweight="bold")

plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")

plt.grid(True, linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

### Observation

- The loss decreases steadily as the number of epochs increases.
- This indicates that the Gradient Descent algorithm successfully minimizes the error and converges toward the optimal solution.
- The convergence demonstrates that the learning rate is appropriate for training the model.

## 10.8 : Regression Line using Gradient Descent 

```python
plt.figure(figsize=(9,5))

plt.scatter(
    X_test_simple,
    y_test_simple,
    color="#EC4899",
    alpha=0.7,
    s=45,
    label="Actual Data"
)

plt.plot(
    X_test_simple,
    y_pred_gd,
    color="#7C3AED",
    linewidth=2.5,
    label="Gradient Descent Line"
)

plt.title("Gradient Descent Regression Line",
          fontsize=16,
          fontweight="bold")

plt.xlabel("Area")
plt.ylabel("House Price")

plt.grid(True, linestyle="--", alpha=0.3)

plt.legend()

plt.tight_layout()
plt.show()
```

## 10.9 : Compare with Scikit-Learn 

```python
print("Comparison with Scikit-Learn")
print("-"*50)

print(f"Scikit-Learn Slope      : {simple_model.coef_[0]:.4f}")
print(f"Gradient Descent Slope  : {m:.4f}")

print()

print(f"Scikit-Learn Intercept      : {simple_model.intercept_:.4f}")
print(f"Gradient Descent Intercept  : {c:.4f}")
```

### Observation

# Step 12: Multiple Linear Regression using Gradient Descent (From Scratch)

In this step, Gradient Descent is implemented from scratch for **Multiple Linear Regression**. Unlike Simple Linear Regression, this model learns multiple weights simultaneously, where each weight corresponds to an input feature. The objective is to minimize the Mean Squared Error (MSE) loss function through iterative parameter updates.

## 12.1 : Prepare the Training Data

```python
X_train = X_train_multiple.values
X_test = X_test_multiple.values

y_train = y_train_multiple.values
y_test = y_test_multiple.values
```

## 12.2 : Initialize Parameters 

```python
# Number of features
n_features = X_train.shape[1]

# Initialize weights and bias
weights = np.zeros(n_features)
bias = 0

learning_rate = 1e-10
epochs = 1000

n = len(y_train)

loss_history_multi = []
```

## 12.3 : Implement Gradient Descent

```python
for epoch in range(epochs):

    # Prediction
    y_pred = np.dot(X_train, weights) + bias

    # Loss
    loss = np.mean((y_train - y_pred) ** 2)
    loss_history_multi.append(loss)

    # Gradients
    dw = (-2/n) * np.dot(X_train.T, (y_train - y_pred))
    db = (-2/n) * np.sum(y_train - y_pred)

    # Update
    weights = weights - learning_rate * dw
    bias = bias - learning_rate * db
```

## 12.4 : Display Learned Parameters 

```python
weights_df = pd.DataFrame({
    "Feature": X_multiple.columns,
    "Gradient Descent Weight": weights
})

print("Bias (Intercept):", bias)

weights_df
```

## 12.5 : Make Predictions

```python
y_pred_multi_gd = np.dot(X_test, weights) + bias

print("First 5 Predictions:")
print(y_pred_multi_gd[:5])
```

## 12.6 : Evaluate the Model

```python
mae_multi_gd = mean_absolute_error(y_test, y_pred_multi_gd)
mse_multi_gd = mean_squared_error(y_test, y_pred_multi_gd)
rmse_multi_gd = np.sqrt(mse_multi_gd)
r2_multi_gd = r2_score(y_test, y_pred_multi_gd)

print("Mean Absolute Error:", mae_multi_gd)
print("Mean Squared Error:", mse_multi_gd)
print("Root Mean Squared Error:", rmse_multi_gd)
print("R² Score:", r2_multi_gd)
```

## 12.7 : Loss vs Epoch

```python
plt.figure(figsize=(9,5))

plt.plot(
    loss_history_multi,
    color="#7C3AED",
    linewidth=2.5
)

plt.title("Multiple Linear Regression - Loss vs Epoch",
          fontsize=16,
          fontweight="bold")

plt.xlabel("Epoch")
plt.ylabel("Mean Squared Error")

plt.grid(True, linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

### Observation

- The training loss decreases continuously with each epoch, indicating successful optimization.
- The Gradient Descent algorithm converges as the loss stabilizes.
- This demonstrates that the model successfully learns the optimal weights for multiple input features.

## 12.8 : Actual vs Predicted

```python
plt.figure(figsize=(8,6))

plt.scatter(
    y_test,
    y_pred_multi_gd,
    color="#EC4899",
    alpha=0.7,
    s=50
)

plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    color="#7C3AED",
    linewidth=2
)

plt.title("Actual vs Predicted (Gradient Descent)",
          fontsize=16,
          fontweight="bold")

plt.xlabel("Actual House Price")
plt.ylabel("Predicted House Price")

plt.grid(True, linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```
